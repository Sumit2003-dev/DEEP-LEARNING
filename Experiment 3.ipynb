### **3. Mathematical Foundations**

#### **3.1 Single-Layer Perceptron**
A perceptron computes a weighted sum of inputs plus a bias, passing the scalar result through a threshold step function:

$$z = \mathbf{w}^T \mathbf{x} + b = \sum_{i=1}^n w_i x_i + b$$

$$\hat{y} = f(z) = \begin{cases} 1 & \text{if } z \ge 0 \\ 0 & \text{if } z < 0 \end{cases}$$

**Weight Update Rule**:
For learning rate $\alpha$ and true target $y \in \{0, 1\}$:
$$w_i \leftarrow w_i + \alpha (y - \hat{y}) x_i$$
$$b \leftarrow b + \alpha (y - \hat{y})$$

#### **3.2 Multi-Layer Perceptron (2-Layer Neural Network Architecture)**

```
Input Layer (X)           Hidden Layer (A^[1])            Output Layer (A^[2])
  [x1] ───────────────────► ( a1^[1] ) ───────────────────► ( y_hat )
        \   W^[1], b^[1]  /            \   W^[2], b^[2]  /
  [x2] ───────────────────► ( a2^[1] ) ───────────────────►
        \                /              \                /
         ────────────────► ( a3^[1] ) ───►
```

#### **Forward Propagation Equations (Vectorized across $m$ samples):**
1. **Hidden Layer (Layer 1)**:
   $$\mathbf{Z}^{[1]} = \mathbf{X} \mathbf{W}^{[1]} + \mathbf{b}^{[1]} \quad \in \mathbb{R}^{m \times n_h}$$
   $$\mathbf{A}^{[1]} = \text{ReLU}(\mathbf{Z}^{[1]}) = \max(0, \mathbf{Z}^{[1]}) \quad \text{or} \quad \tanh(\mathbf{Z}^{[1]})$$
2. **Output Layer (Layer 2)**:
   $$\mathbf{Z}^{[2]} = \mathbf{A}^{[1]} \mathbf{W}^{[2]} + \mathbf{b}^{[2]} \quad \in \mathbb{R}^{m \times 1}$$
   $$\mathbf{A}^{[2]} = \sigma(\mathbf{Z}^{[2]}) = \frac{1}{1 + e^{-\mathbf{Z}^{[2]}}} \quad \in \mathbb{R}^{m \times 1}$$
3. **Binary Cross-Entropy Loss**:
   $$\mathcal{L} = -\frac{1}{m} \sum_{i=1}^m \left[ y^{(i)} \ln(A^{[2](i)}) + (1 - y^{(i)}) \ln(1 - A^{[2](i)}) \right]$$

#### **Backward Propagation Equations (Chain Rule Derivation):**
1. **Output Layer Gradients**:
   $$d\mathbf{Z}^{[2]} = \frac{\partial \mathcal{L}}{\partial \mathbf{Z}^{[2]}} = \mathbf{A}^{[2]} - \mathbf{Y} \quad \in \mathbb{R}^{m \times 1}$$
   $$d\mathbf{W}^{[2]} = \frac{\partial \mathcal{L}}{\partial \mathbf{W}^{[2]}} = \frac{1}{m} (\mathbf{A}^{[1]})^T d\mathbf{Z}^{[2]} \quad \in \mathbb{R}^{n_h \times 1}$$
   $$d\mathbf{b}^{[2]} = \frac{\partial \mathcal{L}}{\partial \mathbf{b}^{[2]}} = \frac{1}{m} \sum_{i=1}^m d\mathbf{Z}^{[2]} \quad \in \mathbb{R}^{1 \times 1}$$

2. **Hidden Layer Gradients**:
   $$d\mathbf{A}^{[1]} = d\mathbf{Z}^{[2]} (\mathbf{W}^{[2]})^T \quad \in \mathbb{R}^{m \times n_h}$$
   $$d\mathbf{Z}^{[1]} = d\mathbf{A}^{[1]} \odot g'(\mathbf{Z}^{[1]}) \quad (\text{where } \odot \text{ is Hadamard element-wise product})$$
   $$\text{For ReLU: } g'(\mathbf{Z}^{[1]}) = \mathbf{1}(\mathbf{Z}^{[1]} > 0)$$
   $$\text{For Tanh: } g'(\mathbf{Z}^{[1]}) = 1 - (\mathbf{A}^{[1]})^2$$
   $$d\mathbf{W}^{[1]} = \frac{1}{m} \mathbf{X}^T d\mathbf{Z}^{[1]} \quad \in \mathbb{R}^{n_x \times n_h}$$
   $$d\mathbf{b}^{[1]} = \frac{1}{m} \sum_{i=1}^m d\mathbf{Z}^{[1]} \quad \in \mathbb{R}^{1 \times n_h}$$

3. **Parameter Updates (Gradient Descent)**:
   $$\mathbf{W}^{[l]} \leftarrow \mathbf{W}^{[l]} - \alpha \, d\mathbf{W}^{[l]}$$
   $$\mathbf{b}^{[l]} \leftarrow \mathbf{b}^{[l]} - \alpha \, d\mathbf{b}^{[l]}$$

### **4. Step-by-Step Implementation**

#### **Step 1: Imports & Setup**


````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons

# Set random seed for exact reproducibility
np.random.seed(42)

print("[SUCCESS] NumPy and Matplotlib initialized!")
````

#### **Step 2: Building the Single-Layer Perceptron Class from Scratch**


````python
class PerceptronFromScratch:
    """
    Single-Layer Rosenblatt Perceptron classifier.
    """
    def __init__(self, learning_rate=0.1, max_epochs=50):
        self.lr = learning_rate
        self.epochs = max_epochs
        self.weights = None
        self.bias = 0.0
        self.errors_history = []

    def _activation(self, z):
        # Heaviside Step Function
        return np.where(z >= 0.0, 1, 0)

    def fit(self, X, y):
        n_samples, n_features = X.shape
        # Initialize weights to zeros (or small random values)
        self.weights = np.zeros(n_features)
        self.bias = 0.0
        self.errors_history = []

        for epoch in range(self.epochs):
            total_error = 0
            for xi, target in zip(X, y):
                # Forward linear combination
                linear_output = np.dot(xi, self.weights) + self.bias
                prediction = self._activation(linear_output)
                
                # Compute error
                error = target - prediction
                total_error += abs(error)
                
                # Weight & Bias Update
                self.weights += self.lr * error * xi
                self.bias += self.lr * error

            self.errors_history.append(total_error)
            if total_error == 0:
                print(f"[CONVERGED] Converged at epoch {epoch + 1}!")
                break
        return self

    def predict(self, X):
        linear_output = np.dot(X, self.weights) + self.bias
        return self._activation(linear_output)
````

#### **Step 3: Training Perceptron on Linearly Separable Gates (AND, OR)**


````python
# Define Boolean Logic Datasets
X_logic = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])

y_and = np.array([0, 0, 0, 1])  # AND Gate
y_or  = np.array([0, 1, 1, 1])  # OR Gate
y_xor = np.array([0, 1, 1, 0])  # XOR Gate (Non-linear)

print("=== 1. TRAINING ON AND GATE ===")
p_and = PerceptronFromScratch(learning_rate=0.2, max_epochs=20)
p_and.fit(X_logic, y_and)
preds_and = p_and.predict(X_logic)
print(f"Weights: {p_and.weights}, Bias: {p_and.bias}")
print(f"Predictions: {preds_and} | Targets: {y_and} -> Accuracy: {np.mean(preds_and == y_and)*100:.0f}%
")

print("=== 2. TRAINING ON OR GATE ===")
p_or = PerceptronFromScratch(learning_rate=0.2, max_epochs=20)
p_or.fit(X_logic, y_or)
preds_or = p_or.predict(X_logic)
print(f"Weights: {p_or.weights}, Bias: {p_or.bias}")
print(f"Predictions: {preds_or} | Targets: {y_or} -> Accuracy: {np.mean(preds_or == y_or)*100:.0f}%
")
````

#### **Step 4: Demonstrating Perceptron Failure on XOR (The Non-Linear Problem)**
In 1969, Marvin Minsky and Seymour Papert proved that a single-layer perceptron can only learn **linearly separable** functions. For an XOR gate, no single straight line can separate `(0, 1)` and `(1, 0)` from `(0, 0)` and `(1, 1)`. As a result, the perceptron never converges.


````python
print("=== 3. TRAINING ON XOR GATE (EXPECTING FAILURE) ===")
p_xor = PerceptronFromScratch(learning_rate=0.2, max_epochs=20)
p_xor.fit(X_logic, y_xor)
preds_xor = p_xor.predict(X_logic)
print(f"Weights: {p_xor.weights}, Bias: {p_xor.bias}")
print(f"Predictions: {preds_xor} | Targets: {y_xor} -> Accuracy: {np.mean(preds_xor == y_xor)*100:.0f}%")
print("[OBSERVATION] The single-layer perceptron failed to classify XOR correctly!")

# Plotting the Linear Separability comparison
fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))

def plot_logic_gate(ax, X, y, title, weights=None, bias=None):
    for i, label in enumerate(y):
        marker = 'o' if label == 1 else 's'
        color = '#2b5c8f' if label == 1 else '#e05d44'
        ax.scatter(X[i, 0], X[i, 1], color=color, s=150, marker=marker, edgecolors='black', zorder=5)
    ax.set_xlim(-0.5, 1.5)
    ax.set_ylim(-0.5, 1.5)
    ax.set_title(title, fontsize=12, fontweight='bold')
    ax.set_xlabel("Input X1")
    ax.set_ylabel("Input X2")
    ax.grid(True, alpha=0.3)
    
    # Plot decision boundary line if weights exist: w1*x1 + w2*x2 + b = 0 => x2 = -(w1*x1 + b)/w2
    if weights is not None and weights[1] != 0:
        x_vals = np.linspace(-0.5, 1.5, 100)
        y_vals = -(weights[0] * x_vals + bias) / weights[1]
        ax.plot(x_vals, y_vals, color='green', lw=2, linestyle='--', label='Decision Boundary')
        ax.legend()

plot_logic_gate(axes[0], X_logic, y_and, "AND Gate (Linearly Separable)", p_and.weights, p_and.bias)
plot_logic_gate(axes[1], X_logic, y_or,  "OR Gate (Linearly Separable)", p_or.weights, p_or.bias)
plot_logic_gate(axes[2], X_logic, y_xor, "XOR Gate (Non-Linearly Separable)", None, None)

plt.tight_layout()
plt.show()
````

#### **Step 5: Implementing a Multi-Layer Neural Network Class in Pure NumPy**
To overcome the linear limitation, we introduce **Hidden Layers** with non-linear activation functions.
We build `NeuralNetworkFromScratch` implementing:
- Flexible hidden layer size
- Forward propagation (Tanh hidden activation, Sigmoid output activation)
- Analytical backpropagation using matrix calculus
- Gradient descent optimizer


````python
class NeuralNetworkFromScratch:
    """
    A 2-Layer Fully-Connected Artificial Neural Network from scratch in NumPy.
    Architecture: Input (n_x) -> Hidden (n_h with Tanh) -> Output (1 with Sigmoid)
    """
    def __init__(self, n_x, n_h, n_y=1, learning_rate=0.1, seed=42):
        np.random.seed(seed)
        self.lr = learning_rate
        
        # Xavier / Glorot Initialization for weights
        self.W1 = np.random.randn(n_x, n_h) * np.sqrt(2.0 / (n_x + n_h))
        self.b1 = np.zeros((1, n_h))
        self.W2 = np.random.randn(n_h, n_y) * np.sqrt(2.0 / (n_h + n_y))
        self.b2 = np.zeros((1, n_y))
        
        self.loss_history = []

    # Activation Functions & Derivatives
    @staticmethod
    def _sigmoid(z):
        return 1.0 / (1.0 + np.exp(-np.clip(z, -500, 500)))

    @staticmethod
    def _tanh(z):
        return np.tanh(z)

    @staticmethod
    def _tanh_derivative(a):
        # a is already tanh(z)
        return 1.0 - a ** 2

    def forward(self, X):
        """Forward pass through hidden and output layers."""
        self.Z1 = np.dot(X, self.W1) + self.b1           # Shape: (m, n_h)
        self.A1 = self._tanh(self.Z1)                    # Shape: (m, n_h)
        self.Z2 = np.dot(self.A1, self.W2) + self.b2     # Shape: (m, n_y)
        self.A2 = self._sigmoid(self.Z2)                 # Shape: (m, n_y)
        return self.A2

    def compute_loss(self, y_true, y_pred):
        """Binary Cross-Entropy Loss."""
        m = y_true.shape[0]
        # Numerical epsilon to prevent log(0)
        eps = 1e-15
        y_pred = np.clip(y_pred, eps, 1.0 - eps)
        loss = - (1.0 / m) * np.sum(y_true * np.log(y_pred) + (1.0 - y_true) * np.log(1.0 - y_pred))
        return loss

    def backward(self, X, y):
        """Backpropagation to compute analytical gradients."""
        m = X.shape[0]

        # Output Layer Gradients
        dZ2 = self.A2 - y                                # Shape: (m, n_y)
        dW2 = (1.0 / m) * np.dot(self.A1.T, dZ2)         # Shape: (n_h, n_y)
        db2 = (1.0 / m) * np.sum(dZ2, axis=0, keepdims=True)  # Shape: (1, n_y)

        # Hidden Layer Gradients
        dA1 = np.dot(dZ2, self.W2.T)                     # Shape: (m, n_h)
        dZ1 = dA1 * self._tanh_derivative(self.A1)       # Shape: (m, n_h)
        dW1 = (1.0 / m) * np.dot(X.T, dZ1)              # Shape: (n_x, n_h)
        db1 = (1.0 / m) * np.sum(dZ1, axis=0, keepdims=True)  # Shape: (1, n_h)

        # Gradient Descent Updates
        self.W2 -= self.lr * dW2
        self.b2 -= self.lr * db2
        self.W1 -= self.lr * dW1
        self.b1 -= self.lr * db1

    def fit(self, X, y, epochs=5000, print_every=1000):
        y = y.reshape(-1, 1)
        self.loss_history = []

        for epoch in range(1, epochs + 1):
            y_pred = self.forward(X)
            loss = self.compute_loss(y, y_pred)
            self.loss_history.append(loss)
            self.backward(X, y)

            if epoch % print_every == 0 or epoch == 1:
                preds = (y_pred >= 0.5).astype(int)
                acc = np.mean(preds == y) * 100
                print(f"Epoch {epoch:5d}/{epochs} | Loss: {loss:.5f} | Accuracy: {acc:.1f}%")
        return self

    def predict(self, X):
        prob = self.forward(X)
        return (prob >= 0.5).astype(int)

    def predict_proba(self, X):
        return self.forward(X)
````

#### **Step 6: Solving the XOR Problem with the Multi-Layer Neural Network**


````python
# XOR Dataset
X_xor = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y_xor = np.array([[0], [1], [1], [0]])

print("=== TRAINING MULTI-LAYER NEURAL NETWORK ON XOR ===")
# Architecture: 2 inputs -> 4 hidden neurons (tanh) -> 1 output (sigmoid)
mlp_xor = NeuralNetworkFromScratch(n_x=2, n_h=4, n_y=1, learning_rate=0.8, seed=42)
mlp_xor.fit(X_xor, y_xor, epochs=3000, print_every=1000)

xor_preds = mlp_xor.predict(X_xor)
xor_probs = mlp_xor.predict_proba(X_xor)

print("
--- FINAL XOR EVALUATION ---")
for i in range(4):
    print(f"Input: {X_xor[i]} => Target: {y_xor[i, 0]} | Predicted Prob: {xor_probs[i, 0]:.4f} => Class: {xor_preds[i, 0]}")

print(f"
[VERIFICATION] Multi-Layer Neural Network XOR Accuracy: {np.mean(xor_preds == y_xor)*100:.0f}%!")
````

#### **Step 7: Training on a Non-Linear 2D Dataset (`make_moons`)**
To test generalization and non-linear manifold mapping, we generate interleaving half-circles (`make_moons`).


````python
# Generate Synthetic Non-Linear Moons Dataset
X_moons, y_moons = make_moons(n_samples=400, noise=0.20, random_state=42)
y_moons = y_moons.reshape(-1, 1)

print(f"Moons Dataset generated: X = {X_moons.shape}, y = {y_moons.shape}")

# Train Neural Network with 8 hidden units
mlp_moons = NeuralNetworkFromScratch(n_x=2, n_h=8, n_y=1, learning_rate=0.3, seed=42)
print("
--- TRAINING ON MOONS DATASET ---")
mlp_moons.fit(X_moons, y_moons, epochs=4000, print_every=1000)

moons_acc = np.mean(mlp_moons.predict(X_moons) == y_moons) * 100
print(f"
Final Moons Training Accuracy: {moons_acc:.2f}%")
````

#### **Step 8: Visualizing the Learned Non-Linear Decision Boundary & Loss Curve**


````python
fig, axes = plt.subplots(1, 2, figsize=(15, 5.5))

# 1. Loss Trajectory Plot
axes[0].plot(mlp_moons.loss_history, color='#2b5c8f', lw=2)
axes[0].set_title("Training Loss Convergence Curve (Binary Cross-Entropy)", fontsize=12, fontweight='bold')
axes[0].set_xlabel("Epochs")
axes[0].set_ylabel("Loss")
axes[0].grid(True, alpha=0.3)

# 2. Decision Boundary Contour Plot
x_min, x_max = X_moons[:, 0].min() - 0.5, X_moons[:, 0].max() + 0.5
y_min, y_max = X_moons[:, 1].min() - 0.5, X_moons[:, 1].max() + 0.5
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 200), np.linspace(y_min, y_max, 200))

# Forward pass on dense grid
grid_points = np.c_[xx.ravel(), yy.ravel()]
grid_probs = mlp_moons.predict_proba(grid_points).reshape(xx.shape)

# Plot contour
contour = axes[1].contourf(xx, yy, grid_probs, levels=25, cmap='RdBu', alpha=0.7)
plt.colorbar(contour, ax=axes[1], label='Class 1 Probability')

# Scatter actual data points
scatter = axes[1].scatter(
    X_moons[:, 0], X_moons[:, 1],
    c=y_moons.ravel(), cmap=plt.cm.coolwarm,
    edgecolors='k', s=40, zorder=5
)
axes[1].set_title(f"Learned Non-Linear Decision Boundary (Accuracy: {moons_acc:.1f}%)", fontsize=12, fontweight='bold')
axes[1].set_xlabel("Feature 1")
axes[1].set_ylabel("Feature 2")

plt.tight_layout()
plt.show()
````

### **5. Theoretical Discussion & Architectural Insights**

1. **Why Single Perceptrons Cannot Solve XOR**:
   - The decision boundary of a single neuron is given by $\mathbf{w}^T \mathbf{x} + b = 0$, which is geometrically a flat hyperplane (a straight line in 2D).
   - In the XOR problem, the outputs are:
     - $(0, 0) \to 0$
     - $(1, 1) \to 0$
     - $(0, 1) \to 1$
     - $(1, 0) \to 1$
   - It is geometrically impossible to draw a single straight line that isolates the diagonal coordinates $(0, 1)$ and $(1, 0)$ from $(0, 0)$ and $(1, 1)$.

2. **How Hidden Layers and Non-Linear Activations Solve XOR**:
   - The hidden layer maps the input space $(x_1, x_2)$ into a higher-dimensional feature space $(a_1^{[1]}, a_2^{[1]})$.
   - Because the activation function ($\tanh$ or $\text{ReLU}$) is non-linear, it "bends" and warps the coordinate space such that the data becomes linearly separable by the output neuron.
   - If linear activations were used ($g(z) = z$), composition of linear functions remains strictly linear ($\mathbf{W}^{[2]}(\mathbf{W}^{[1]}\mathbf{X}) = \mathbf{W}_{\text{comb}} \mathbf{X}$), making multiple layers completely redundant!

3. **Weight Initialization (Symmetry Breaking)**:
   - If all weights are initialized to zeros or identical constants, every hidden neuron computes identical outputs and receives identical backpropagation gradients ($dW$).
   - This "symmetry" prevents neurons from specializing in distinct features. Random initialization (Xavier/He) breaks this symmetry.

### **6. Viva-Voce / Review Questions with Answers**

**Q1: Why is an activation function essential in hidden layers of a neural network?**
* **Answer**: Without non-linear activation functions, regardless of how many layers are stacked, the network is merely computing a linear combination of linear transformations ($\mathbf{W}_2 \mathbf{W}_1 \mathbf{x} = \mathbf{W}_{net} \mathbf{x}$). Non-linear activations enable neural networks to act as **Universal Function Approximators**, allowing them to model arbitrary complex non-linear decision boundaries.

**Q2: What is the XOR problem, and who mathematically demonstrated the limitation of single-layer perceptrons?**
* **Answer**: The XOR (Exclusive OR) problem is a non-linearly separable classification task. In 1969, Marvin Minsky and Seymour Papert published the book *Perceptrons*, proving mathematically that single-layer perceptrons cannot compute the XOR function, initiating the first "AI Winter" until multi-layer backpropagation was popularized.

**Q3: What happens if all weights in a multi-layer neural network are initialized to zero?**
* **Answer**: If weights are initialized to zero, every neuron in a given hidden layer computes identical activations ($0$). During backpropagation, all neurons receive identical gradients and update identically. The network fails to learn diverse representations because the neurons never break symmetry.

**Q4: What is the role of the Chain Rule in Backpropagation?**
* **Answer**: The Chain Rule of calculus enables the calculation of the derivative of composite functions. Backpropagation calculates the rate of change of the global loss function with respect to weights located deep inside the network by chaining local partial derivatives from the output layer backwards: $\frac{\partial \mathcal{L}}{\partial \mathbf{W}^{[1]}} = \frac{\partial \mathcal{L}}{\partial \mathbf{Z}^{[2]}} \cdot \frac{\partial \mathbf{Z}^{[2]}}{\partial \mathbf{A}^{[1]}} \cdot \frac{\partial \mathbf{A}^{[1]}}{\partial \mathbf{Z}^{[1]}} \cdot \frac{\partial \mathbf{Z}^{[1]}}{\partial \mathbf{W}^{[1]}}$.

**Q5: What is Xavier (Glorot) Initialization and why is it preferred over arbitrary normal distribution?**
* **Answer**: Xavier initialization samples weights from a Gaussian distribution with variance $\text{Var}(W) = \frac{2}{n_{in} + n_{out}}$. This ensures that the variance of the activations during the forward pass and the variance of the gradients during the backward pass remain approximately constant across layers, preventing signals from exploding or vanishing in deep networks.
